## Сравнение скорости работы numpy с кодом python

```python
import timeit
import numpy as np
```

```python
npdata = np.array(np.arange(0, 1000000.0))
listdata = list(np.arange(0, 1000000.0))

def sum():
    s = 0;
    for n in listdata:
        s += n
    return s

print(np.sum(npdata))
print(sum())

t1 = timeit.timeit(lambda: np.sum(npdata), number = 100)
print(f"numpy time: {t1}")

t2 = timeit.timeit(lambda: sum() , number = 100)
print(f"python time: {t2}")

print(f"python time / numpy time: {t2 / t1}")
```

## Срезы

Общая форма: X[\<obj\>], где obj – list, tuple, slice (срез), генератор и др

```python
m = np.array([
    [11, 12, 13],
    [21, 22, 23],
    [31, 32, 33]
])

print(m)
```

### Срез матрицы

```python
print("Срез по строке (первая ось):")
print(m[1])
print("")
print("Срез по столбцу (вторая оси):")
print(m[:,1])
print("")
print("Матрица (строки 0,2, столбцы 0-2):")
print(m[[0,2],0:3])
```

### Срез по маске

```python
x = np.array([1, 5, -3, 10, 2])
mask = x > 0
print(mask)
y = x[mask]
print(y)
```
