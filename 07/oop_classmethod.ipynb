# classmethod

## Проблема

```python
class A:
    def func(cls):
        print(f"Текущий класс: {cls}")
A.func(A)
a = A()
a.func(A)
```

## Решение с использованием classmethod

```python
class A:
    def func(cls):
        print(f"Текущий класс: {cls}")
    func = classmethod(func)
A.func()
a = A()
a.func()
```

## Решение с использованием декоратора @classmethod

```python
class A:
    @classmethod
    def func(cls):
        print(f"Текущий класс: {cls}")
A.func()
a = A()
a.func()
```
