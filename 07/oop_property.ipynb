# Свойства (properties)

Атрибуты объекта.

Доступ:
* Только для чтения
* Только для записи
* Чтение-запись

Тип:
* Хранение значения
* Вычисляемое значение (только для чтения)


```python
help(property)
```

# Использование property

```python
class AstroObject:
    def __init__(self, name):
        self.__name = name
    def __getname(self):
        return self.__name
    def __setname(self, n):
        if not isinstance(n, str):
            raise Exception("ERROR")
        self.__name = n
    name = property(__getname, __setname)
```

```python
a = AstroObject("aaa")
a.name = "hello"
print(a.name)
```

# Использование декоратора @property

```python
class A:
    def __init__(self, n):
        self.__name = n
    @property
    def name(self):
        return self.__name
    @name.setter
    def name(self, n):
        self.__name = n
```

```python
a = A("aaa")
print(a.name)
a.name = "bbb"
print(a.name)
```
