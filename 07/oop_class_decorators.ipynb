# Декораторы классов

## Пример: Уникальный идентификатор объекта

```python
def oid(cls):
    class Decorated(cls):
        lastid = 0
        def __init__(self, *args, **kws):
            cls.__init__(self, *args, **kws)
            self.__id = Decorated.lastid
            Decorated.lastid += 1
        def id(self):
            return self.__id
    return Decorated
```

```python
@oid
class A:
    pass
```

```python
print(A)
```

```python
l = [A() for i in range(10)]
print([x.id() for x in l])
```
