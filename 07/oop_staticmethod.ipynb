# staticmethod

## Проблема

```python
class A:
    def func():
        print("A.func")
A.func()
a = A()
a.func()
```

## Решение через staticmethod

```python
class A:
    def func():
        print("A.func")
    func = staticmethod(func)
A.func()
a = A()
a.func()
```

## Решение через декоратор @staticmethod

```python
class A:
    @staticmethod
    def func():
        print("A.func")
A.func()
a = A()
a.func()
```
