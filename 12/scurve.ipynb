```python
from sympy import *
```

```python
# Временная переменная
t = symbols('t')
H = symbols('H')
T = symbols('T')

# Ограничения
jm = symbols('j_m')

# Начальные условия
a0 = symbols("a_0")
v0 = symbols('v_0')

# Входные параметры
#ta = symbols('t_a') # время разгона
tv = symbols('t_v') # время движения с постоянной скоростью
#td = symbols('t_d') # время замедления

ta = together((T - tv) / 2)
td = together((T - tv) / 2)

# Расчет участка разгона
a1_i = a0 + jm * t
v1_i = v0 + integrate(a1_i, t)
h1_i = integrate(v1_i, t)
h1   = h1_i.subs({t:ta})

display(h1)

a2_i = a1_i.subs({t:ta})
v2_i = v1_i.subs({t:ta}) + integrate(a2_i, t)
h2_i = integrate(v2_i, t)
h2   = h2_i.subs({t:tv})

a3_i = a2_i - jm*t
v3_i = v2_i.subs({t:tv}) + integrate(a3_i, t)
h3_i = integrate(v3_i, t)
h3   = h3_i.subs({t:td})

e = simplify(H - h1 - h2 - h3)
display(e)

solveset(e, tv)

# e = h - h1 - h2 - h3
# e = simplify(e)
# solveset(e, ta)
```
